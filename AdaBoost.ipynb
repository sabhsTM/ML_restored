**Z-Score & Outlier Detection** : -
Z-score helps to understand if a data value is greater or smaller than mean and how far away it is from the mean.More specifically, Z score tells how many standard deviations away a data point is from the mean

```python
threshold = 3
from scipy.stats import zscore
numerical_features = ['Fare','Age']
z_score = np.abs(zscore(df[numerical_features]))
print(z_score)
```

```python
outliers = np.where(z_score > threshold)
print(outliers)
```

```python
from scipy.stats.mstats import winsorize
df['Age'] = winsorize(df['Age'],limits = [0.15,0.15])
df['Fare'] = winsorize(df['Fare'],limits = [0.15,0.15])
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder()
df['Sex'] = oe.fit_transform(df[['Sex']])
```

```python
df['Embarked'] = oe.fit_transform(df[['Embarked']])
```

```python
df['Survived'] = oe.fit_transform(df[['Survived']])
```

```python
 df
```

```python
X= df.drop('Survived',axis =1)
y= df['Survived']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=0)
```

```python
from sklearn.ensemble import AdaBoostClassifier
ab= AdaBoostClassifier(n_estimators=100,learning_rate=1,random_state=0)
ab.fit(X_train,y_train)

y_pred = ab.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
accuracy_score(y_pred,y_test)
confusion_matrix(y_pred,y_test)
```
